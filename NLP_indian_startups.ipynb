```python
# ==========================================
# REMOVE UNNECESSARY INDEX COLUMN
# ==========================================

df_indian = df_indian.drop(
    columns=['Unnamed: 0']
)

print("Unnecessary index column removed.")
print(df_indian.columns.tolist())
```

```python
# ==========================================
# CREATE SEARCHABLE STARTUP TEXT
# ==========================================

df_indian['search_text'] = (
    df_indian['Industries'].fillna('') + ' ' +
    df_indian['Description'].fillna('')
)

print("Searchable startup text created.")

df_indian[
    ['Company', 'Industries', 'Description', 'search_text']
].head()
```

```python
# ==========================================
# TF-IDF FOR INDIAN STARTUP SIMILARITY
# ==========================================

from sklearn.feature_extraction.text import TfidfVectorizer

indian_tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words='english',
    ngram_range=(1, 2),
    max_features=10000,
    sublinear_tf=True
)

indian_matrix = indian_tfidf.fit_transform(
    df_indian['search_text']
)

print("TF-IDF vectorization completed.")
print("Matrix shape:", indian_matrix.shape)
```

```python
# ==========================================
# INDIAN STARTUP SIMILARITY FUNCTION
# ==========================================

from sklearn.metrics.pairwise import cosine_similarity

def find_similar_indian_startups(
    startup_description,
    top_n=5
):
    
    # Convert user description into TF-IDF
    query_vector = indian_tfidf.transform(
        [startup_description]
    )
    
    # Compare with Indian startup profiles
    similarity_scores = cosine_similarity(
        query_vector,
        indian_matrix
    ).flatten()
    
    # Get top matching startups
    top_indices = similarity_scores.argsort()[::-1][:top_n]
    
    results = df_indian.iloc[
        top_indices
    ].copy()
    
    # Add similarity score
    results['similarity_score'] = (
        similarity_scores[top_indices]
    )
    
    return results.reset_index(drop=True)
```

```python
# ==========================================
# TEST INDIAN STARTUP RETRIEVAL
# ==========================================

test_description = """
An online platform that connects customers with
local home service professionals such as electricians,
plumbers, cleaners and appliance repair specialists.
"""

similar_indian = find_similar_indian_startups(
    test_description,
    top_n=5
)

similar_indian[
    [
        'Company',
        'City',
        'Industries',
        'Funding Amount in $',
        'Funding Round',
        'No. of Employees',
        'similarity_score'
    ]
]
```

```python
# ==========================================
# DISPLAY SIMILAR INDIAN STARTUPS
# ==========================================

for index, row in similar_indian.iterrows():

    print("\n" + "=" * 60)

    print("Company:", row['Company'])
    print("City:", row['City'])
    print("Starting Year:", row['Starting Year'])
    print("Founders:", row['Founders'])
    print("Industries:", row['Industries'])
    print("Employees:", row['No. of Employees'])
    print("Funding:", row['Funding Amount in $'])
    print("Funding Rounds:", row['Funding Round'])
    print("Investors:", row['No. of Investors'])
    print("Similarity:", round(row['similarity_score'], 4))

    print("\nDescription:")
    print(row['Description'])
```

```python
# ==========================================
# IMPORT SEMANTIC EMBEDDING MODEL
# ==========================================

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

print("Loading semantic embedding model...")

embedding_model = SentenceTransformer(
    'all-MiniLM-L6-v2'
)

print("Embedding model loaded successfully!")
```

```python
# ==========================================
# CREATE SEMANTIC EMBEDDINGS
# ==========================================

indian_embeddings = embedding_model.encode(
    df_indian['search_text'].tolist(),
    show_progress_bar=True
)

print("Embeddings created successfully!")
print("Embedding shape:", indian_embeddings.shape)
```

```python
# ==========================================
# SEMANTIC STARTUP SIMILARITY FUNCTION
# ==========================================

def find_similar_indian_startups_semantic(
    startup_description,
    top_n=5
):
    
    # Convert user's startup idea into embedding
    query_embedding = embedding_model.encode(
        [startup_description]
    )
    
    # Calculate semantic similarity
    similarity_scores = cosine_similarity(
        query_embedding,
        indian_embeddings
    ).flatten()
    
    # Get top matching startups
    top_indices = similarity_scores.argsort()[::-1][:top_n]
    
    results = df_indian.iloc[
        top_indices
    ].copy()
    
    # Add similarity score
    results['similarity_score'] = (
        similarity_scores[top_indices]
    )
    
    return results.reset_index(drop=True)
```

```python
# ==========================================
# TEST SEMANTIC INDIAN STARTUP RETRIEVAL
# ==========================================

test_description = """
An online platform that connects customers with
local home service professionals such as electricians,
plumbers, cleaners and appliance repair specialists.
"""

similar_indian_semantic = find_similar_indian_startups_semantic(
    test_description,
    top_n=5
)

similar_indian_semantic[
    [
        'Company',
        'City',
        'Industries',
        'Funding Amount in $',
        'Funding Round',
        'No. of Employees',
        'similarity_score'
    ]
]
```

```python
# ==========================================
# DISPLAY SEMANTICALLY SIMILAR INDIAN STARTUPS
# ==========================================

for index, row in similar_indian_semantic.iterrows():

    print("\n" + "=" * 60)

    print("Company:", row['Company'])
    print("City:", row['City'])
    print("Starting Year:", row['Starting Year'])
    print("Founders:", row['Founders'])
    print("Industries:", row['Industries'])
    print("Employees:", row['No. of Employees'])
    print("Funding:", row['Funding Amount in $'])
    print("Funding Rounds:", row['Funding Round'])
    print("Investors:", row['No. of Investors'])
    print("Similarity:", round(row['similarity_score'], 4))

    print("\nDescription:")
    print(row['Description'])
```

```python
# ==========================================
# SAVE INDIAN STARTUP SEMANTIC COMPONENT
# ==========================================

import joblib
import os

models_path = r"C:\Users\shubham kharb\Desktop\Minor Project\models"

os.makedirs(models_path, exist_ok=True)

# Save the semantic embedding model
embedding_model.save(
    os.path.join(
        models_path,
        "indian_startup_embedding_model"
    )
)

# Save startup embeddings
joblib.dump(
    indian_embeddings,
    os.path.join(
        models_path,
        "indian_startup_embeddings.pkl"
    )
)

# Save startup knowledge base
joblib.dump(
    df_indian,
    os.path.join(
        models_path,
        "indian_startup_knowledge_base.pkl"
    )
)

print("Indian startup semantic component saved successfully!")
```
